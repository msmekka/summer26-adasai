# What is ADAS?

**ADAS** = Advanced Driver Assistance Systems

Technology that helps drivers by taking over small tasks automatically.

| Feature | What it does |
|---------|-------------|
| Lane Keeping Assist | Steers the car back if you drift out of your lane |
| Automatic Emergency Braking | Stops the car if it detects an imminent collision |
| Adaptive Cruise Control | Maintains a safe following distance automatically |
| Parking Assist | Steers into a parking spot while you control the pedals |
| Blind Spot Warning | Alerts you when a car is in your blind spot |

> 💬 **Discussion: How many of you have been in a car with any of these features? What did it feel like?**

# SAE Levels of Autonomy

The Society of Automotive Engineers defined 6 levels of driving automation.

| Level | Name | Who drives? | Example |
|-------|------|-------------|----------|
| 0 | No Automation | Human always | Regular car |
| 1 | Driver Assistance | Human + one feature | Cruise control |
| 2 | Partial Automation | Human must supervise | Tesla Autopilot |
| 3 | Conditional Automation | Car drives, human on standby | Mercedes Drive Pilot |
| 4 | High Automation | Car handles most situations | Waymo robotaxi |
| 5 | Full Automation | No human ever needed | Does not exist yet |

<br>

<div style='background-color:#D5E8F0; padding:15px; border-radius:8px;'>
<strong>Where is your robot?</strong> By Friday it will be somewhere between Level 3 and Level 4 -- it will drive itself and make AI decisions, but you will supervise it.
</div>

<br>

> 💬 **Discussion: What would need to be true for Level 5 to exist? What problems still need to be solved?**

# How Does a Self-Driving Car See?

Self-driving cars use multiple sensors at the same time -- called **sensor fusion**.

| Sensor | What it detects | Range | Your robot has it? |
|--------|----------------|-------|-------------------|
| Camera | Color, shape, text, faces | Far | ✅ Yes |
| Ultrasonic | Distance to nearby objects | Short | ✅ Yes |
| Radar | Speed and distance of objects | Far | ❌ No |
| Lidar | 3D map of surroundings | Far | ❌ No |
| GPS | Location on a map | Global | ❌ No |

<br>

<div style='background-color:#E2EFDA; padding:15px; border-radius:8px;'>
<strong>Your robot has 2 of these 5 sensors.</strong> Real self-driving cars combine all of them to build a complete picture of the world. More sensors = harder to fool.
</div>

# How We Work This Week

<h2 style='color:white; text-align:center;'>The Tweak-Test-Iterate Loop</h2>
<p style='color:#AED6F1; text-align:center; font-size:20px;'>
Change one thing &nbsp;→&nbsp; Run it &nbsp;→&nbsp; Watch what happens &nbsp;→&nbsp; Repeat

<br>

**The rules:**
- You will NOT write code from scratch
- You WILL change values inside clearly marked zones
- Change ONE thing at a time
- If something breaks -- restart the kernel and start over
- Every experiment teaches you something even if it fails

<br>

> 💬 **This is how real engineers work. No one writes perfect code the first time.**
