# Notebook 01: LED Control

## ADAS Connection
Before a car can drive itself it needs to **communicate**. Modern vehicles use lights to signal intent -- brake lights, turn signals, hazard lights. In autonomous vehicles these signals become even more important because pedestrians and other drivers need to understand what the car is about to do.

Some self-driving car prototypes use **colored LED arrays** to communicate their state: green means moving, red means stopped, yellow means uncertain. You are about to program exactly that.

---

## How It Works
Your robot has an **RGB LED** -- a single light that can produce any color by mixing Red, Green, and Blue values. Each channel is either ON or OFF.

By combining the three channels you get 7 colors plus off:

| Red | Green | Blue | Color |
|-----|-------|------|-------|
| ON  | OFF   | OFF  | Red |
| OFF | ON    | OFF  | Green |
| OFF | OFF   | ON   | Blue |
| ON  | ON    | OFF  | Yellow |
| ON  | OFF   | ON   | Magenta |
| OFF | ON    | ON   | Cyan |
| ON  | ON    | ON   | White |
| OFF | OFF   | OFF  | Off |

---

## The Code
Run this cell first to set up the LED pins.

```python
import RPi.GPIO as GPIO
import time
import signal
import sys

# Pin definitions (BCM numbering)
LED_R = 22   # wiringPi 3
LED_G = 27   # wiringPi 2
LED_B = 24   # wiringPi 5

GPIO.setmode(GPIO.BCM)
GPIO.setwarnings(False)
GPIO.setup(LED_R, GPIO.OUT)
GPIO.setup(LED_G, GPIO.OUT)
GPIO.setup(LED_B, GPIO.OUT)

def set_color(red, green, blue):
    GPIO.output(LED_R, GPIO.HIGH if red   else GPIO.LOW)
    GPIO.output(LED_G, GPIO.HIGH if green else GPIO.LOW)
    GPIO.output(LED_B, GPIO.HIGH if blue  else GPIO.LOW)

def led_off():
    set_color(False, False, False)

print('LED ready!')
```
