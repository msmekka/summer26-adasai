# Notebook 02: Motor Control

## ADAS Connection
Every self-driving car needs **drive-by-wire** -- the ability for a computer to control the engine, brakes, and steering without a human touching the pedals. In this notebook you are writing the code that does exactly that for your robot.

Modern ADAS features like **adaptive cruise control** and **automatic emergency braking** work by sending speed commands to the motor controller -- the same thing you are about to do.

---

## How It Works
Your robot has two motors -- one for the left track and one for the right track. By controlling the speed and direction of each motor independently you can make the robot go forward, backward, turn, or spin in place.

Speed is controlled using **PWM (Pulse Width Modulation)** -- a way of rapidly switching a signal on and off to simulate different power levels. A value of 0 means stopped, 100 means full speed.

## YOUR TURN -- Tweak Zone 2: Turning

Change the turn duration below. How long does it take to make a 90 degree turn at your current speed?

```python
# ═══════════════════════════════════════
#   TWEAK THESE VALUES
TURN_SPEED    = 40      # speed during turn
TURN_DURATION = 0.5     # seconds -- try 0.3, 0.5, 0.8, 1.0
DIRECTION     = 'left'  # 'left' or 'right'
# ═══════════════════════════════════════

print(f'Turning {DIRECTION} for {TURN_DURATION} seconds at speed {TURN_SPEED}...')
if DIRECTION == 'left':
    motors.turn_left(TURN_SPEED)
elif DIRECTION == 'right':
    motors.turn_right(TURN_SPEED)
else:
    print('ERROR: DIRECTION must be left or right')
    
time.sleep(TURN_DURATION)
motors.brake()
print('Done.')
```

## YOUR TURN -- Tweak Zone 3: Drive a Pattern

Now combine forward and turning to drive a pattern. Can you make the robot drive a square?

A square needs 4 forward segments and 4 turns. Tweak the values until it closes back to where it started.

> **Think like an engineer:** This is called **dead reckoning** -- estimating position based on speed and time. Real self-driving cars use this as a backup when GPS fails.

```python
# ═══════════════════════════════════════
#   TWEAK THESE VALUES
DRIVE_SPEED    = 40     # forward speed
DRIVE_DURATION = 1.0    # seconds per side
TURN_SPEED     = 40     # turning speed  
TURN_DURATION  = 0.5    # seconds per turn -- adjust until turn is ~90 degrees
# ═══════════════════════════════════════

print('Driving a square...')
for side in range(4):
    print(f'  Side {side + 1}...')
    motors.forward(DRIVE_SPEED)
    time.sleep(DRIVE_DURATION)
    motors.brake()
    time.sleep(0.2)
    motors.turn_right(TURN_SPEED)
    time.sleep(TURN_DURATION)
    motors.brake()
    time.sleep(0.2)

print('Square complete!')
```
